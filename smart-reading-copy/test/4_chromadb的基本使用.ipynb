```python
from langchain_community.embeddings import DashScopeEmbeddings
from langchain_chroma import Chroma
from langchain_core.documents import Document
```

```python
# 1.创建 Chroma 客户端
# 第一次运行：创建新的 collection
# 第二次运行：自动加载已有的 collection
chroma_db = Chroma(
    # 存放位置
    persist_directory="./chroma_db",
    # 集合等价于数据库中的表
    collection_name="test-collection",
    # 嵌入模型
    embedding_function=DashScopeEmbeddings(),
    # 余弦相似度
    collection_metadata={"hnsw:space":"cosine"}
)
```

```python
# 2. 新增 document
data = [
    Document(page_content="猫坐在垫子上"),
    Document(page_content="一只猫在垫子上休息"),
    Document(page_content="今天天气很好")
]

chroma_db.aadd_documents(
    documents=data,
    ids=[f"id-{i}" for i in range(1, len(data) + 1)]
)

"""
id 原始文本 向量
id-1 猫坐在垫子上 embedding_function.embed_query("猫坐在垫子上")
id-2 一只猫在垫子上休息 embedding_function.embed_query("一只猫在垫子上休息")
id-3 今天天气很好 embedding_function.embed_query("今天天气很好")
"""
```

```python
# 3.相似度查询（和mysql的where等值查询不一样，我们这里是基于相似度匹配）
user_query = "猫在哪？"
# result = chroma_db.similarity_search(user_query, k=2)
result = chroma_db.similarity_search_with_relevance_scores(user_query, k=2)
print(type(result))
print(result)
```
